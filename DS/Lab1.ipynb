### Визначення номера варіанта
Для вибору варіанта використовується оператор:
$$N = \text{ord}(X) \pmod{10} + 1$$
де $X$ - перша літера імені в латинській транскрипції.

- Перша літера імені: **'D'**
- Код символу: `ord('D') = 68`
- Розрахунок: $68 \pmod{10} + 1 = 8 + 1 = 9$
- **Номер варіанта:** **9**

**Завдання за варіантом 9:**
1. **Завдання 3:** Чеська республіка (оформлення за допомогою Markdown та HTML тегів).
2. **Завдання 4:** У введеному списку залишити тільки непрості числа і відсортувати їх за зменшенням.

```python
# Розрахунок номера варіанта
first_letter = 'D'
variant = ord(first_letter) % 10 + 1
print(f"First letter: '{first_letter}' (ASCII: {ord(first_letter)})")
print(f"variant: {variant}")
```

# Завдання 4 (Варіант 9)

**Умова:**
> У введеному списку залишити тільки **непрості числа** і відсортувати їх **за зменшенням**.
> Організувати контроль значень, що вводяться (якщо у списку присутнє нечислове значення, повертати повідомлення: `"Помилка. У списку присутнє нечислове значення"`).

**Математична основа:**
- Число називається **складеним (непростим)**, якщо воно ціле, строго більше 1 і має щонайменше один дільник, відмінний від 1 та самого себе.
- Числа $0$ та $1$ за математичним означенням **не є ані простими, ані складеними** (тому вони не потрапляють у результуючий список, що повністю узгоджується з контрольними прикладами у завданні).
- Від'ємні числа також не є простими чи складеними натуральними числами.

```python
def is_prime(n: int) -> bool:
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True


def task_9(raw_input_data=None):
    if raw_input_data is None:
        raw_input_data = input("Enter integers separated by space: ")

    if isinstance(raw_input_data, str):
        tokens = raw_input_data.strip().split()
    elif isinstance(raw_input_data, (list, tuple)):
        tokens = [str(x) for x in raw_input_data]
    else:
        return "Помилка. У списку присутнє нечислове значення"

    nums = []
    for token in tokens:
        try:
            nums.append(int(token))
        except ValueError:
            return "Помилка. У списку присутнє нечислове значення"

    composites = [x for x in nums if not is_prime(x)]
    composites.sort(reverse=True)
    return composites
```
